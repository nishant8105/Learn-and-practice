```python
X_encode1 = pd.get_dummies(X, columns=['model', 'transmission', 'fuelType'], drop_first=True)
X_encode1
```

```python
X_encode1 = X_encode1.astype(int)
X_encode1
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
columns=['model', 'transmission', 'fuelType']
X_encode2 = X.copy()
label_encode = {}

for col in columns:
    le = LabelEncoder()
    X_encode2[col] = le.fit_transform(X_encode2[col].astype(str))
    label_encode[col] = le
```

```python
X_encode2
```

```python
X_encode2['model'].value_counts()
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numerical_col = ['year', 'mileage', 'tax', 'mpg', 'engineSize']
X_encode1[numerical_col]= scaler.fit_transform(X_encode1[numerical_col])
```

```python
X_encode1
```

```python
columns = ['model', 'year', 'transmission', 'mileage', 'fuelType', 'tax', 'mpg','engineSize']
X_encode2[columns] = scaler.fit_transform(X_encode2[columns])
```

```python
X_encode2
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encode1, y, test_size=0.20, random_state=42)
```

```python
model = LinearRegression()
model.fit(X_train, y_train)
```

```python
ypred = model.predict(X_test)
ypred
```

```python
r2 = r2_score(y_test, ypred)
r2
```

```python
n = X_test.shape[0]
p = X_test.shape[1]

adjusted_r2 = 1 - ((1 - r2)*(n - 1))/(n-p-1)
print("adjusted_r2", adjusted_r2)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X_encode2, y, test_size=0.20, random_state=42)
```

```python
model2 = LinearRegression()
model2.fit(X_train, y_train)
```

```python
ypred = model2.predict(X_test)
ypred
```
